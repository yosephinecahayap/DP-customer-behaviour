# **Customer Behaviour Analysis**

Dataset Customer Shopping Trends Indian diperoleh dari Kaggle (URL: https://www.kaggle.com/datasets/rewantbhriguvanshi/customer-shopping-trends-indian ). Dataset ini berisi data transaksi pembelian pelanggan dalam konteks retail dan e-commerce di India selama periode Januari 2023 hingga Desember 2024.

Dataset mencakup berbagai informasi terkait pelanggan, transaksi, produk, metode pembayaran, promosi, pengiriman, dan pengalaman pelanggan. Informasi tersebut dapat digunakan untuk memahami pola pembelian, preferensi pelanggan, serta faktor-faktor yang berkaitan dengan nilai transaksi dan kepuasan pelanggan.

Dataset ini terdiri dari 25 kolom yang mencakup beberapa kelompok informasi utama:
- Informasi Pelanggan: usia, jenis kelamin, lokasi, status berlangganan, dan frekuensi pembelian.
- Informasi Transaksi: ID transaksi, ID pelanggan, tanggal pembelian, kanal pembelian, jumlah barang, dan jumlah pembelian.
- Informasi Produk: kategori produk, nama barang, merek, warna, dan ukuran.
- Promosi dan Pembayaran: diskon, acara promosi, biaya pengiriman, dan metode pembayaran.
- Pengalaman Pelanggan: waktu pengiriman, rating pelanggan, status pengembalian barang, dan jumlah pembelian sebelumnya.

## Purchase Analysis

```python
print("Total Purchase Amount:", df['purchase_amount'].sum())
print("Average Purchase Amount:", df['purchase_amount'].mean())
print("Median Purchase Amount:", df['purchase_amount'].median())
```

Dataset mencatat total nilai pembelian sebesar ₹64.676.945 dari 10.000 transaksi. Rata-rata nilai pembelian per transaksi sebesar ₹6.467,69, sedangkan mediannya sebesar ₹3.491,00. Perbedaan antara rata-rata dan median menunjukkan bahwa terdapat transaksi dengan nilai pembelian relatif tinggi yang memengaruhi rata-rata keseluruhan.

```python
channel_summary = df.groupby('purchase_channel').agg(
    transactions=('transaction_id', 'nunique'),
    customers=('customer_id', 'nunique'),
    total_purchase=('purchase_amount', 'sum'),
    avg_purchase=('purchase_amount', 'mean')
).reset_index()

print(channel_summary)
```

```python
plt.figure(figsize=(7, 4))
bars = plt.bar(
    channel_summary['purchase_channel'],
    channel_summary['total_purchase']
)
plt.title('Total Purchase by Channel')
plt.xlabel('Purchase Channel')
plt.ylabel('Total Purchase (₹)')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{int(bar.get_height()):,}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

Transaksi online mendominasi dengan 7.417 transaksi (74,17%), dibandingkan transaksi offline sebanyak 2.583 transaksi (25,83%). Total nilai pembelian online mencapai ₹47.817.573, lebih tinggi daripada offline sebesar ₹16.859.372. Namun, rata-rata nilai pembelian offline sedikit lebih tinggi, yaitu ₹6.527,05 dibandingkan online sebesar ₹6.447,02.

Hasil ini menunjukkan bahwa kanal online mendominasi dari segi jumlah transaksi dan total nilai pembelian, sedangkan rata-rata nilai pembelian per transaksi relatif serupa pada kedua kanal.

## Customer Analysis

One-Time vs Repeat Customers

```python
customer_orders = df.groupby('customer_id')['transaction_id'].nunique()

one_time = (customer_orders == 1).sum()
repeat_cust = (customer_orders > 1).sum()

print("Total customer:", len(customer_orders))
print("Jumlah customer yang melakukan transaksi satu kali:", one_time)
print("Jumlah customer yang melakukan transaksi lebih dari satu kali:", repeat_cust)
```

```python
customer_type = ['One-Time Customers', 'Repeat Customers']
customer_count = [one_time, repeat_cust]

plt.figure(figsize=(7, 4))
bars = plt.bar(customer_type, customer_count)

plt.title('One-Time vs Repeat Customers')
plt.xlabel('Customer Type')
plt.ylabel('Number of Customers')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{int(bar.get_height()):,}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

Dari total 3.291 pelanggan, sebanyak 2.693 pelanggan (81,83%) melakukan transaksi lebih dari satu kali, sedangkan 598 pelanggan (18,17%) hanya melakukan satu transaksi. Mayoritas pelanggan dalam dataset ini tercatat melakukan pembelian berulang. Temuan ini dapat menjadi dasar untuk mengeksplorasi pola pembelian pelanggan lebih lanjut, meskipun belum cukup untuk menyimpulkan tingkat loyalitas pelanggan.

Kategori Footwear mencatat total nilai pembelian tertinggi sebesar ₹37.462.463 dari 3.395 transaksi, meskipun jumlah transaksinya lebih sedikit daripada Clothing.
```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[17, 24, 34, 44, 54, 64],
    labels=['18-24', '25-34', '35-44', '45-54', '55-64']
)

age_group_purchase = df.groupby('age_group', observed=True)['purchase_amount'].agg(
    transactions='count',
    total_purchase='sum',
    avg_purchase='mean'
)

print(age_group_purchase)
```

Kelompok usia 55–64 tahun memiliki rata-rata nilai pembelian tertinggi sebesar ₹6.657,74, sedangkan kelompok usia 45–54 tahun memiliki rata-rata terendah sebesar ₹6.136,49. Meskipun terdapat perbedaan, nilai rata-rata pembelian antarkelompok usia relatif berdekatan. Pada dataset ini, kelompok usia tidak menunjukkan perbedaan rata-rata nilai pembelian yang terlalu besar.

### Promotion Behavior

```python
print("Average Discount:", df['discount'].mean())
print("Median Discount:", df['discount'].median())
print("Maximum Discount:", df['discount'].max())
```

```python
df['discount_group'] = pd.cut(
    df['discount'],
    bins=[-1, 10, 20, 30, 40, 50, 100],
    labels=['0-10%', '11-20%', '21-30%', '31-40%', '41-50%', '>50%']
)

discount_group_purchase = df.groupby('discount_group', observed=True)['purchase_amount'].agg(
    transactions='count',
    total_purchase='sum',
    avg_purchase='mean'
)

print(discount_group_purchase)
```

```python
plt.figure(figsize=(8, 4))
bars = plt.bar(
    discount_group_purchase.index.astype(str),
    discount_group_purchase['transactions']
)

plt.title('Transactions by Discount Range')
plt.xlabel('Discount Range')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=30)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{int(bar.get_height()):,}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

Rentang diskon 0–10% mendominasi dengan 6.693 transaksi, disusul rentang 11–20% sebanyak 2.702 transaksi. Rata-rata diskon keseluruhan sebesar 8,70%, dengan median 8% dan nilai maksimum 69%. Jumlah transaksi pada rentang diskon di atas 20% jauh lebih sedikit. Hal ini menunjukkan bahwa transaksi dalam dataset lebih banyak terjadi pada rentang diskon yang relatif rendah.

```python
plt.figure(figsize=(8, 4))
bars = plt.bar(
    discount_group_purchase.index.astype(str),
    discount_group_purchase['avg_purchase']
)

plt.title('Average Purchase by Discount Range')
plt.xlabel('Discount Range')
plt.ylabel('Average Purchase (₹)')
plt.xticks(rotation=30)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'₹{bar.get_height():,.0f}',
        ha='center',
        va='bottom',
        fontsize=8
    )

plt.tight_layout()
plt.show()
```

Rentang diskon 41–50% memiliki rata-rata nilai pembelian tertinggi sebesar ₹8.615,67 per transaksi. Namun, rentang tersebut hanya mencakup 61 transaksi. Sementara itu, rentang 0–10% memiliki rata-rata ₹6.448,03 dari 6.693 transaksi. Nilai rata-rata pembelian tidak meningkat secara konsisten seiring bertambahnya diskon. Karena jumlah transaksi pada beberapa rentang diskon cukup sedikit, hasil ini tidak dapat dijadikan bukti bahwa diskon tinggi menyebabkan nilai pembelian meningkat.
